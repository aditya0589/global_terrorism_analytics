## Temporal trends of global Terrorism from 1970 to 2017

```sql
%sql
SELECT COUNT(*) FROM terrorism_final;
```

```sql
SELECT year, COUNT(*) as count_attacks FROM terrorism_final
GROUP BY year
ORDER BY count_attacks DESC
LIMIT 20;
```

```sql
%python
import plotly.express as px

query = """
SELECT year, COUNT(*) as count_attacks FROM terrorism_final
GROUP BY year
ORDER BY year ASC
"""

df = spark.sql(query)
pdf = df.toPandas()

fig = px.line(pdf, x='year', y='count_attacks', title='Temporal Trends of Terrorist Attack Counts (1970-2017)')
fig.update_traces(mode='lines+markers')
fig.show()
```

We can find that the number of terrorist incidents have been growing from sharply till 2014. But since then, the number of incidents have been reduced.

The sharp rise in the number of incidents from 2011 to 2014. This was the time during the syrian civil war and the rise of ISIL and the Arab spring. In 2014, the United States under the Barak Obama Administration launched airstrikes against ISIS in Syria and Iraq followed by a multinational coalition which eventually overthrew ISIS

```sql
SELECT year, SUM(casualties) AS total_casualties
FROM terrorism_final
GROUP BY year
ORDER BY total_casualties DESC
LIMIT 20    
```

```sql
%python
import plotly.express as px

query = """
SELECT year, SUM(casualties) AS total_casualties
FROM terrorism_final
GROUP BY year
ORDER BY year ASC
"""

df = spark.sql(query)
pdf = df.toPandas()

fig = px.line(pdf, x='year', y='total_casualties', title='Temporal Trends of Total Casualties (1970-2017)')
fig.update_traces(mode='lines+markers')
fig.show()
```

We find the number of casualties have sharply reduced since 2014. This can be attributed to the decrease in the total number of attacks during these years

```sql
SELECT month,
       COUNT(*) AS total_attacks,
       ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM terrorism_final), 2) AS pct
FROM terrorism_final
WHERE month != 0
GROUP BY month
ORDER BY month;
```


The monthly data reveals notable seasonality in terrorist attacks. Certain months, such as May, June, and July, tend to have higher frequencies of incidents, while others like February and December see fewer attacks. This pattern may reflect geopolitical events, weather conditions, or operational strategies of terrorist groups. Understanding these trends can help inform prevention and response strategies.

```sql
SELECT
    CASE
        WHEN year BETWEEN 1970 AND 1979 THEN '1970s'
        WHEN year BETWEEN 1980 AND 1989 THEN '1980s'
        WHEN year BETWEEN 1990 AND 1999 THEN '1990s'
        WHEN year BETWEEN 2000 AND 2009 THEN '2000s'
        ELSE '2010s'
    END AS decade,
    COUNT(*) AS total_attacks,
    SUM(COALESCE(casualties, 0)) AS total_killed,
    ROUND(AVG(COALESCE(casualties, 0)), 2) AS avg_killed_per_attack
FROM terrorism_final
GROUP BY decade
ORDER BY decade;
```

We find that the post 2010 world experienced the most number of terrorist attacks. The possible reasons might be the rise of ISIS, the afghanistan conflict and the Iraq instability.
However, the 2000s faced the highest casualties per attack. This suggests more lethal attacks during this period than in the 2010s
