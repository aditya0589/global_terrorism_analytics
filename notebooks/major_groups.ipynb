```sql
SELECT DISTINCT
    gname
FROM terrorism_final;
```

```sql
SELECT gname, COUNT(*) AS attacks
FROM terrorism_final
GROUP BY gname
ORDER BY attacks DESC
LIMIT 10;
```

```sql
WITH top_groups AS (
  SELECT gname
  FROM terrorism_final
  GROUP BY gname
  ORDER BY COUNT(*) DESC
  LIMIT 10
)
SELECT
  t.gname,
  FLOOR(t.year / 10) * 10 AS decade,
  COUNT(*) AS attacks
FROM terrorism_final t
JOIN top_groups tg ON t.gname = tg.gname
GROUP BY t.gname, FLOOR(t.year / 10) * 10
ORDER BY t.gname, decade
```

## Middle East and North Africa

```sql
SELECT
    gname,
    COUNT(*) AS attacks
FROM terrorism_final
WHERE region = "Middle East & North Africa" AND gname != "Unknown"
GROUP BY gname
ORDER BY attacks DESC
LIMIT 10;
```
